```python
import pandas as pd
import numpy as np
import os
```

```python
basic_df = pd.read_csv('../../collected_data/basic.csv')
ouid_df = pd.read_csv('../../collected_data/ouid.csv')
```

```python
# 매너 등급 수치화

manner_score_map = {
    "최고": 3,
    "매우 좋음": 2,
    "좋음": 1,
    "보통": 0,
    "우려": -1,
    "위험": -2,
    "나쁨": -3
}

# 계정 생성 날짜 yyyy-mm-dd 형식으로 변환
basic_df['user_date_create'] = pd.to_datetime(basic_df['user_date_create']).dt.date

# manner 컬럼을 수치로 변환
basic_df["manner_score"] = basic_df["manner_grade"].map(manner_score_map)

# 클랜이 있으면 1 없으면 0으로
basic_df['clan_flag'] = np.where(basic_df['clan_name'].isnull(), 0, 1)

# 불필요한 컬럼, 점수 변환한 컬럼은 제거
basic_df = basic_df.drop(['title_name','manner_grade','clan_name'], axis=1)
```

```python
basic_df.copy().to_csv('../../collected_data/basic_final.csv', index=False, encoding='utf-8-sig')
```
